```python
import pandas as pd
import seaborn
from IPython.display import display
import numpy as np

# Excelファイルを読み込む（ファイル名とシート名を適宜変更）
# 例：ファイル名 'data.xlsx'、シート名 'Sheet1'
df_Conflict = pd.read_excel('scene1_2nd.xlsx', sheet_name='葛藤', usecols='A:D')

# データ（625行×4列）を前提とする
# 各125行が1つの水準に対応
levels = [
    "葛藤なし×連続",
    "葛藤なし×同時",
    "葛藤あり×連続",
    "葛藤あり×同時",
]
n_subjects = df_Conflict.shape[0] // len(levels)  # 行数を水準数で割って被験者数を計算
n_levels = len(levels)
n_items = df_Conflict.shape[1]  # 列数を自動取得

# 水準・被験者情報追加
level_labels = sum([[lvl]*n_subjects for lvl in levels], [])
subject_ids = list(range(1, n_subjects+1)) * n_levels

df_long = df_Conflict.copy()
df_long['level'] = level_labels
df_long['subject'] = subject_ids

# 項目Q1〜Q4を縦持ちに
df_long = pd.melt(
    df_long,
    id_vars=['subject', 'level'],
    value_vars=df_Conflict.columns.tolist(),  # ← ここを修正
    var_name='item',
    value_name='score'
)


# 項目平均をとって水準ごとの得点に
df_mean = df_long.groupby(['subject', 'level'])['score'].mean().reset_index()
# level と subject をカテゴリ変数に変換（C()なしで済む）
df_mean['level'] = df_mean['level'].astype('category')
df_mean['subject'] = df_mean['subject'].astype('category')
df_wide = df_mean.pivot(index='subject', columns='level', values='score')

conflict = df_wide.copy()
```
